# Nancora Quickstart Notebook

This notebook demonstrates dataset exploration and target-aware analysis using **Nancora** (`nancora`).

```python
import pandas as pd
import nancora as nc

# Print installed version
print("Nancora Version:", nc.__version__)
```

## 1. Create Sample Dataset

```python
df = pd.DataFrame({
    "age": [22, 25, 47, 52, 46, 56, 48, 33, 27, 29],
    "income": [25000, 32000, 75000, 89000, 71000, 95000, 82000, 49000, 39000, 42000],
    "city": ["Surat", "Surat", "Mumbai", "Delhi", "Mumbai", "Delhi", "Surat", "Mumbai", "Delhi", "Surat"],
    "purchased": ["no", "no", "yes", "yes", "yes", "yes", "yes", "no", "no", "no"],
})
df
```

## 2. Unsupervised Exploration

Evaluating `result` in a Jupyter notebook cell renders rich HTML recommendations output!

```python
result = nc.explore(df)
result
```

## 3. Target-Aware Analysis

```python
target_result = nc.explore(df, target="purchased")
target_result
```
